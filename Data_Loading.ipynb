```python
#csv file using read and write
import pandas as pd

# 1. Sample Data Dataframe banao
data = {
    'Name': ['Amit', 'Priya', 'Rahul'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Mathura', 'Agra']
}
df = pd.DataFrame(data)

# 2. DataFrame ko CSV file me save karo (Write)
file_name = "name.csv"
df.to_csv(file_name, index=False)
print(f"File '{file_name}' successfully create ho gayi!")

# 3. Created CSV file ko padho (Read)
df_read = pd.read_csv(file_name)
print("\nFile se padha gaya Data:")
print(df_read)
```

EXCEL FILE CONVERT :-

```python
# 1. Sample DataFrame

data = {
    'Name': ['Amit', 'Priya', 'Rahul'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Mathura', 'Agra']
}
df = pd.DataFrame(data)
# 2. DataFrame ko Excel file (.xlsx) me write karo

excel_file = "name.xlsx"
df.to_excel(excel_file, index=False, sheet_name='Sheet1')

print(f"File '{excel_file}' successfully create ho gayi!")

# 3. Created Excel file ko read karke verify karo
df_excel = pd.read_excel(excel_file)
print("\nExcel File se padha gaya Data:")
print(df_excel)
```

JSON FILE CONVERT :-

```python
# 1. Sample DataFrame
data = {
    'Name': ['Amit', 'Priya', 'Rahul'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Mathura', 'Agra']
}
df = pd.DataFrame(data)

# 2. DataFrame ko JSON file me write (save) karo
json_file = "name.json"
df.to_json(json_file, orient='records', indent=4)

print(f"File '{json_file}' successfully create ho gayi!")

# 3. Created JSON file ko read karke verify karo
df_json = pd.read_json(json_file)
print("\nJSON File se padha gaya Data:")
print(df_json)
```

TAB FILE CONVERT :-

```python
# 1. Sample DataFrame
data = {
    'Name': ['Amit', 'Priya', 'Rahul'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Mathura', 'Agra']
}
df = pd.DataFrame(data)

# 2. DataFrame ko Tab-Separated file (.tsv ya .txt) me save karo
tsv_file = "name.tsv"
df.to_csv(tsv_file, sep='\t', index=False)

print(f"File '{tsv_file}' successfully create ho gayi!")

# 3. Created TSV file ko read karke verify karo
df_tsv = pd.read_csv(tsv_file, sep='\t')
print("\nTSV File se padha gaya Data:")
print(df_tsv)
```

FWF FILE CONVERT :-

```python
# 1. Sample DataFrame
data = {
    'Name': ['Amit', 'Priya', 'Rahul'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Mathura', 'Agra']
}
df = pd.DataFrame(data)

# 2. DataFrame ko FWF file (.txt ya .fwf) me save karo
fwf_file = "name.fwf"
with open(fwf_file, "w") as f:
    f.write(df.to_string(index=False))

print(f"File '{fwf_file}' successfully create ho gayi!")

# 3. Created FWF file ko read karke verify karo
df_fwf = pd.read_fwf(fwf_file)
print("\nFWF File se padha gaya Data:")
print(df_fwf)
```

USE OF nrows :-

```python
df = pd.read_csv("name.csv",sep =",",nrows=2)
print(df)
```

USE of usecols :-

```python
df = pd.read_csv("name.csv",sep =",",usecols=["Age"])
print(df)
```

USE OF index_col

```python
df = pd.read_csv("name.csv",sep =",",index_col="Age")
print(df)
```

Use of skiprows :-

```python
df = pd.read_csv("name.csv",sep =",",skiprows=2)
print(df)
```

USE OF header :-

```python
df = pd.read_csv("name.csv",sep =",",header = None)
print(df)
```
